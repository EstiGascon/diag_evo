## What is a Jupyter Notebook?

A Jupyter Notebook is an interactive document that combines:
- **Code cells** — write and execute Python code
- **Markdown cells** — formatted text, headings, lists, images (like this cell)
- **Output** — results appear directly below the code cell

Notebooks are widely used for data analysis, visualisation, and interactive workflows.

## Cell Types

### Code Cells
- Execute Python code
- Output appears below the cell
- Variables are shared across all code cells in the notebook

### Markdown Cells
- Formatted text (headings, **bold**, *italic*, lists, tables, images)
- Double-click to edit, `Shift+Enter` to render
- Used for documentation and explanations

## Interactive Widgets

Jupyter supports interactive widgets (dropdowns, sliders, maps, buttons).  
The `diag_evo` tool makes extensive use of these for configuration.

Here is a simple example:

````python
import ipywidgets as widgets
from IPython.display import display

# A simple slider widget
slider = widgets.IntSlider(value=5, min=1, max=15, step=1, description='Days ahead:')
display(slider)
````

````python
# The slider value is accessible programmatically
print(f"Current slider value: {slider.value}")
````

## Important: Execution Order Matters

Cells must be run **in order** (top to bottom) for variables to be defined correctly.  
If you skip a cell or run cells out of order, you may get `NameError` or unexpected results.

**Tip:** If things go wrong, restart the kernel and re-run from the top.

````python
# Example: this cell depends on 'x' defined above
# If you run this without running the earlier cell, you'll get an error
print(f"x is still {x} — because variables persist in the kernel's memory")
````

## Useful Tips & Troubleshooting

| Problem | Solution |
|---------|----------|
| Kernel stuck / unresponsive | **Kernel → Restart Kernel** (or stop + start) |
| `NameError: name 'x' is not defined` | Run cells in order from the top |
| Server unresponsive | **File → Hub Control Panel → Stop Server** → refresh → start again |
| Need to install a package | Run `!pip install package_name` in a code cell |

### Keyboard Shortcuts (most useful)

| Shortcut | Action |
|----------|--------|
| `Shift + Enter` | Run cell and move to next |
| `Ctrl + Enter` | Run cell (stay in place) |
| `Esc` then `A` | Insert cell above |
| `Esc` then `B` | Insert cell below |
| `Esc` then `M` | Convert cell to Markdown |
| `Esc` then `Y` | Convert cell to Code |
| `Esc` then `DD` | Delete cell |

## Storage on ECMWF Systems

| Location | Purpose | Persistence |
|----------|---------|-------------|
| `$HOME` | Small files, code repos | Permanent, limited space |
| `$PERM` | Code repositories, notebooks | Permanent, more space |
| `$SCRATCH` | Large data files (GRIB, output) | Temporary (auto-purged) |

**Recommendation:** Keep the `diag_evo` repo on `$PERM` (or `$HOME`) and set the data output directory to `$SCRATCH`.
